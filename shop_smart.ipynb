```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import r2_score,accuracy_score,f1_score
```

```python
# load the dataset
data = pd.read_csv('shop_smart_ecommerce.csv')
data.head()
```

```python
data.columns
```

```python
# Features selection and target variable definition

features = ['ProductRelated', 'ProductRelated_Duration','ExitRates','PageValues', 'SpecialDay', 'Month','VisitorType']
target = "Revenue"
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data['Month']=le.fit_transform(data['Month'])
data['VisitorType']=le.fit_transform(data['VisitorType'])
data['Weekend']=le.fit_transform(data['Weekend'])
data['Revenue']=le.fit_transform(data['Revenue'])
x = data[features]
y = data[target]
```

```python
# train and test split

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=42
)
```

```python
# training model 

model = DecisionTreeClassifier()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print("accuracy: ", accuracy_score(y_test,y_pred))
f1 = f1_score(y_test, y_pred)
print("F1 Score:", f1)
```

Pruning 

```python
# pruning for deapth

deapth=[1,2,3,4,5,6,7,8]
for d in deapth:
    model=DecisionTreeClassifier(max_depth=d)
    model.fit(x_train,y_train)
    acc=model.score(x_test,y_test)
    print(f"accuracy for {d}: ",acc)
```

```python
# pruning for split

splits=[3,6,9,12]

for s in splits:
    model=DecisionTreeClassifier(max_depth=5,min_samples_split=s)
    model.fit(x_train,y_train)
    acc=model.score(x_test,y_test)
    print(f"split for {s}: ",acc)
```

```python
model=DecisionTreeClassifier(max_depth=5,min_samples_split=9)
model.fit(x_train,y_train)
f=f1_score(y_test,y_pred=model.predict(x_test))
print(f)
```
